# 2. 📐 Sistem Koordinat p5.js

Canvas menggunakan sistem koordinat:

```text
(0,0)
  ┌──────────────────────────────→ X
  │
  │
  │
  ↓
  Y
```

- `x` semakin besar → bergerak ke kanan.
- `y` semakin besar → bergerak ke bawah.

Contoh:

```javascript
p.rect(30, 20, 55, 55);
```

Artinya:
- X = 30
- Y = 20
- lebar = 55
- tinggi = 55

# 🟢 LATIHAN TERBIMBING 1 — Matahari

## Tujuan
Memahami cara kerja:
- `circle()`
- `line()`
- koordinat
- ukuran
- warna

### Cara berpikir

Matahari dapat dipecah menjadi:

```text
Matahari
├── circle() → pusat
└── line()   → sinar
```

Ikuti contoh kode berikut terlebih dahulu.

````
%%html
<script src="https://cdnjs.cloudflare.com/ajax/libs/p5.js/1.11.1/p5.min.js"></script>

<div id="contoh_matahari"></div>

<script>
new p5(function(p) {

  p.setup = function() {
    p.createCanvas(800, 500);
  };

  p.draw = function() {
    p.background(235);


    // =====================================
    // CONTOH 1 — MATAHARI
    // =====================================

    p.fill(255, 200, 0);
    p.noStroke();
    p.circle(400, 250, 120);

    p.stroke(255, 180, 0);
    p.strokeWeight(5);

    p.line(400, 160, 400, 100);
    p.line(400, 340, 400, 400);
    p.line(310, 250, 250, 250);
    p.line(490, 250, 550, 250);

    p.line(336, 186, 294, 144);
    p.line(464, 186, 506, 144);
    p.line(336, 314, 294, 356);
    p.line(464, 314, 506, 356);

  };

}, 'contoh_matahari');
</script>
````

### 🔎 Bedah Kode

Perhatikan:

```javascript
p.circle(400, 250, 120);
```

`400` = X  
`250` = Y  
`120` = diameter

Sedangkan:

```javascript
p.line(x1, y1, x2, y2);
```

digunakan untuk menghubungkan dua titik.

### Eksperimen

Sebelum membuat ulang:
1. Ubah posisi matahari.
2. Ubah ukuran matahari.
3. Ubah warna matahari.
4. Tambahkan satu sinar.
